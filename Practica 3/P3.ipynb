1.  **Preprocesado de datos (Vectorización):**
    * Las listas de enteros tienen longitudes variables, lo cual no es admisible para una red densa estándar. Debe transformar las listas en tensores de tamaño fijo.
    * **Tarea:** Implemente una función para realizar **One-hot Encoding** (o Multi-hot) que convierta cada reseña en un vector de dimensión 10,000 (donde 1 indica la presencia de la palabra y 0 su ausencia).
    * Vectorice tanto los datos de train como los de test.

```python
import numpy as np

def one_hot_encoding(data, dimension=10000):
    # Creamos una matriz llena de ceros
    # Filas = número de reseñas. Columnas = numero de palabras (10000)
    results = np.zeros((len(data), dimension))

    # Rellenamos con 1 las posiciones de las palabras que sí aparecen
    for i, data in enumerate(data):
        results[i, data] = 1.

    return results

# Aplicamos la función a nuestros datos
X_train = one_hot_encoding(train_data)
X_test = one_hot_encoding(test_data)

# Imprimimos los resultados para comprobar que ha funcionado
print("Primer dato:", train_data[0])
print("Forma de los datos de entrenamiento (x_train):", X_train.shape)
print("Ejemplo de la primera reseña vectorizada:", X_train[0])
```

Transformamos las respuestas en vectores numéricos

```python
Y_train = np.asarray(train_labels).astype('float32')
Y_test = np.asarray(test_labels).astype('float32')

# Mostramos que sale correcto:
print("Forma de Y_train:", Y_train.shape)
print("Ejemplo de las primeras 5 respuestas:", Y_train[:5])
```

---

2.  **Construcción del Modelo:**
    * Diseñe una red neuronal secuencial (`tf.keras.Sequential`).
    * Use 2 o 3 capas densas (`Dense`).
    * Use funciones de activación `relu` para las capas ocultas.
    * **Importante:** La capa de salida debe tener 1 sola neurona con activación `sigmoid` (para clasificación binaria).

Ponemos 15 y 10 como número de neuronas en cada capa para forzar al modelo a aprender representaciones más generales en vez de memorizar

```python
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

model= Sequential()
model.add( Input(shape=(X_train.shape[1],)))
model.add( Dense (15 , activation = 'relu') )
model.add( Dense (10 , activation = 'relu') )
model.add( Dense (1 , activation = 'sigmoid') )
```

**OPTIMIZADOR**: Vamos a usar el modelo Adaptive Moment Estimation (adam)

```python
from tensorflow.keras.optimizers import Adam

# Tasa de aprendizaje
lmbda= 0.0001
optimizer= Adam(learning_rate= lmbda)
```

**FUNCIÓN DE PERDIDA**: Usamos `binary_crossentropy`

```python
from tensorflow.keras.losses import binary_crossentropy

loss_function= binary_crossentropy
```

**MÉTRICAS**: Solo `accuracy`

```python
metrics= ['accuracy']
```

**COMPILACIÓN MODELOS**: Ya tenemos todo lo necesario para compilarlo y mostrar un resumen del mismo.

```python
model.compile(loss= loss_function, optimizer= optimizer, metrics= metrics)
model.summary()
```

Separamos el 20% de los datos de entrenamiento para validación:

```python
from sklearn.model_selection import train_test_split
X_train_partial, X_train_val, Y_train_partial, Y_train_val  = train_test_split(X_train, Y_train, test_size=0.2, random_state=42, stratify=Y_train) # satratify hace que haya el mismo porcentaje de ceros y unos
```

Podriamos haber puesto `validation_split=0.2` pero Keras no mezcla los datos antes de hacer el corte, por tanto usa las últimas filas que se correspondan con el 20% para validar. El problema es que si está ordenado el dataset, de forma que las positivas fueran las 15000 primeras y las 10000 últimas negativas, solo estaríamos validando con reseñas negativas.

Para solucionarlo he dividido como en las practicas anteriores de forma que haya el mismo porcentaje más o menos para entrenar y validar.

Ahora vamos a entrenar durante 20 épocas:

```python
bs= 128 # Tamaño del batch
epochs= 20 # Épocas de entrenamiento

history = model.fit(X_train_partial, Y_train_partial, batch_size= bs, epochs=epochs, validation_data=(X_train_val, Y_train_val))
```

Los valores del tamaño de batch y de la tasa de aprendizaje se han puesto así porque tras probar con varios valores estos son los que menos overfitting generan, ahora se verá en las gráficas.

4.  **Evaluación:**

En las gráficas se puede observar que hay overfitting a partir de la epoca 7 u 8, donde las curvas de entrenamiento y validación se empiezan a separar cada vez más, ya que la de entrenamiento sigue mejorando pero la de validación o se estanca, como pasa en la precisión, o empieza a tener mayor pérdida.

El modelo obtiene alrededor de `88%` de precisión en el conjunto de test, lo que indica que es capaz de clasificar correctamente la mayoría de reseñas nuevas. El valor de precisión junto con el de pérdida en test de alrededor a `0.3` concuerdan con los obtenidos en la validación anterior (esto puede cambiar en cada ejecución): `val_accuracy: 0.8850` y `val_loss: 0.3148` lo que indica que el conjunto de validación era representativo y el entrenamiento ha sido estable. Con estos resultados podemos decir que el modelo generaliza correctamente a datos no vistos.

## 2.1 Contexto: Eliminación de Ruido en Imágenes (Denoising Autoencoder)

En este ejercicio utilizaremos una arquitectura de tipo **Autoencoder Convolucional** para una tarea de aprendizaje no supervisado: la restauración de imágenes.

El objetivo no es clasificar qué número hay en la imagen, sino aprender a **limpiar** una imagen que ha sido corrompida por ruido aleatorio.

**Dataset:** MNIST (Dígitos manuscritos). Nosotros introduciremos el ruido artificialmente.

```python
# Carga y normalización de MNIST
(x_train, _), (x_test, _) = keras.datasets.mnist.load_data()

# Normalización [0, 1] y reshape a (28, 28, 1)
x_train = x_train.astype('float32') / 255.
x_test = x_test.astype('float32') / 255.
x_train = np.reshape(x_train, (len(x_train), 28, 28, 1))
x_test = np.reshape(x_test, (len(x_test), 28, 28, 1))

# Función para añadir ruido aleatorio
def add_noise(images, noise_factor=0.5):
    noisy_imgs = images + noise_factor * np.random.normal(loc=0.0, scale=1.0, size=images.shape)
    return np.clip(noisy_imgs, 0., 1.)

x_train_noisy = add_noise(x_train)
x_test_noisy = add_noise(x_test)

# Visualización del ruido
n = 5
plt.figure(figsize=(10, 2))
for i in range(n):
    ax = plt.subplot(1, n, i + 1)
    plt.imshow(tf.squeeze(x_test_noisy[i]))
    plt.gray()
    ax.get_xaxis().set_visible(False)
    ax.get_yaxis().set_visible(False)
plt.show()
```

1.  **Construcción del Autoencoder:**
    * Debe construir un modelo que tenga dos partes: **Encoder** y **Decoder**.
    * **Encoder:** Use capas `Conv2D` y `MaxPooling2D` para comprimir la imagen.
    * **Decoder:** Use capas `Conv2D` y `UpSampling2D` (o `Conv2DTranspose`) para reconstruir la imagen al tamaño original (28x28).
    * La capa de salida debe tener 1 filtro y activación `sigmoid` (para que los píxeles estén entre 0 y 1).

**ENCODER**:
- `Conv2D` Aplica filtros para escanear la imagen. Los filtros convolucionales ignoran bastante bien los puntos aleatorios introducidos porque buscan patrones con sentido geométrico como líneas o curvas.

- `MaxPooling2D` Esta capa reduce el tamaño de la imagen a la mitad (de 28x28 a 14x14, y luego a 7x7). Se va quedando con el píxel con el valor más alto de cada zona. Tiene sentido usarlo ya que al aplicar el máximo como el fondo tiene colores con valores cercanos a 0 y el trazo del número tiene valores cercanos a 1, entonces siempre se queda con el del trazo y va eliminando la "suciedad".

**DECODER**:
- `UpSampling2D` imagen de 7x7 píxeles a la que había llegado el Encoder y la "estira" duplicando su tamaño (pasa a 14x14 y luego a 28x28). El problema es que UpSampling clona píxeles, por lo que la imagen estirada queda muy pixelada.

- `Conv2D` en esta fase toma esa imagen estirada y borrosa y aprende a rellenar los huecos y a suavizar bordes pixelados para reconstruir un trazo mas limpio.


```python
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D

autoencoder = Sequential([

    # Imagenes ruidosas de (28x28x1).
    Input(shape=(28, 28, 1)),

    # ENCODER
    ###################################################
    # 32 filtros de 3x3 , funcion de activacion relu, padding='same' (añadira un borde de 0)
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    # Reduce el tamaño a la mitad: la imagen pasa de 28x28 a 14x14
    MaxPooling2D((2, 2), padding='same'),

    ## Sobre la imagen reducida
    # 16 filtros de 3x3 , funcion de activacion relu, padding='same' (añadira un borde de 0)
    Conv2D(16, (3, 3), activation='relu', padding='same'),
    # Reduce otra vez a la mitad: pasa de 14x14 a 7x7
    MaxPooling2D((2, 2), padding='same'),



    # DECODER
    ###################################################
    # Igual que como terminamos el encoder
    Conv2D(16, (3, 3), activation='relu', padding='same'),
    # Duplica el tamaño: la imagen vuelve a subir de 7x7 a 14x14
    UpSampling2D((2, 2)),

    # Sigue perfilando los bordes y formas del número
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    # Duplica el tamaño otra vez: vuelve de 14x14 a 28x28 (Tamaño original)
    UpSampling2D((2, 2)),

    # CAPA DE SALIDA
    ###################################################
    # 1 filtro para fusionar todo en una sola capa de imagen (blanco y negro).
    # 'sigmoid' para obligar a los píxeles a valer entre 0 y 1.
    Conv2D(1, (3, 3), activation='sigmoid', padding='same')
])

# Para ver que todo es correcto
autoencoder.summary()
```

---

2.  **Entrenamiento:**
    * **Entrada (X):** Imágenes con ruido (`x_train_noisy`).
    * **Objetivo (Y):** Imágenes originales limpias (`x_train`).
    * Pérdida: `binary_crossentropy` (funciona bien píxel a píxel) o `mse`.
    * Entrene por al menos 10 épocas.

Usaremos como optimizador adam por ser el más eficiente.

```python
autoencoder.compile(optimizer='adam', loss='binary_crossentropy')

history_autoencoder = autoencoder.fit(
    x=x_train_noisy,          # Imagenes con ruido
    y=x_train,                # Imágenes limpias
    epochs=10,                # Al menos 10 epocas
    batch_size=128,           # Procesa las imágenes en bloques de 128 en 128
    validation_data=(x_test_noisy, x_test) # Para validar
)
```

Se puede ver que no hay overfitting, ya que la pérdida en entrenamiento y en validación disminuye de forma muy similar. Si lo mostráramos con las gráficas como en el ejercicio anterior, la distancia entre ambas lineas se mantendría constante

---

3.  **Resultados:**
    * Utilice el modelo entrenado para limpiar las imágenes de test (`x_test_noisy`).

```python
# Usamos el modelo entrenado para limpiar las imágenes de test
imagenes_reconstruidas = autoencoder.predict(x_test_noisy)

# Mostramos 5 ejemplos
n = 5
plt.figure(figsize=(10, 6))

for i in range(n):
    # IMAGEN ORIGINAL
    ax = plt.subplot(3, n, i + 1)
    plt.imshow(tf.squeeze(x_test[i]), cmap='gray')
    if i == 2: plt.title("IMAGEN ORIGINAL", fontsize=14, pad=10)
    plt.axis('off')

    # IMAGEN RUIDOSA
    ax = plt.subplot(3, n, i + 1 + n)
    plt.imshow(tf.squeeze(x_test_noisy[i]), cmap='gray')
    if i == 2: plt.title("IMAGEN RUIDOSA", fontsize=14, pad=10)
    plt.axis('off')

    # IMAGEN RECONSTRUIDA
    ax = plt.subplot(3, n, i + 1 + 2*n)
    plt.imshow(tf.squeeze(imagenes_reconstruidas[i]), cmap='gray')
    if i == 2: plt.title("IMAGEN RECONSTRUIDA", fontsize=14, pad=10)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

# **Conclusiones Generales**

Añada una celda de texto final comentando las diferencias fundamentales que ha observado en la forma de entrenar un modelo supervisado (donde le damos una etiqueta de clase) frente al no supervisado (donde la propia imagen limpia actúa como supervisor).

En el aprendizaje supervisado el modelo recibe durante el entrenamiento una reseña y su valor 0 o 1. La función de pérdida mide la diferencia entre la predicción y esa etiqueta externa. Para las reseñas necesitamos que una persona las clasifique con 0 y 1 para poder entrenar al modelo.

En el autoencoder en cambio, la entrada y la salida esperada son el mismo tipo de dato, una imagen. El modelo recibe la imagen con ruido como entrada y la imagen limpia como objetivo. No hay etiquetas, el propio dato actúa como supervisor de sí mismo.

La diferencia principal es que en el supervisado el modelo aprende a distinguir entre clases, mientras que en el autoencoder aprende a entender la estructura interna de los datos, qué es información relevante y qué es ruido, sin necesidad de que nadie le indique a qué clase pertenece cada imagen. Esto lo hace especialmente útil cuando no se dispone de datos etiquetados.


Otra diferencia es que aunque usamos la misma función de pérdida `binary_crossentropy`, esta se aplica sobre estructuras completamente distintas. En las reseñas se compara un único valor predicho con una etiqueta, mientras que en el autoencoder se compara cada píxel de la imagen reconstruida con su píxel original.

Por último en el de las reseñas tiene sentido usar `accuracy` porque hay clases correctas e incorrectas y entonces podemos medir el porcentaje de aciertos. Sin embargo en las imágenes no hay correctas o incorrectas, luego no tiene sentido usarlo porque no estamos clasificando, nos vale con la función de perdida, ya que mide cuanto se "aleja" de la imagen real.
